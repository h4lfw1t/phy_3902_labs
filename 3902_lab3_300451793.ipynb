## Abstract

This experiment analyzed a 650nm laser diode with a built-in series resistor and lens with 5 volts of power applied by shining it on a light-dependent resistor (LDR) used to analyze the intensity of the laser's light. The setup was used for three similar experiments to determine the polarization of the beam as well as observe the effects of a linearly polarized filter in the beam path at various angles. The next step was verifying Brewster's angle by rotating a glass slide in the beam path and analyzing the relationship between the angle of incidence into the slide and the light intensity. The final step was achieving a Gaussian bell curve for the shape and intensity distribution of a beam using a blade to slowly uncover the beam and graphing the gradient of the intensity readings as the blade moved. All the results were as expected; it was determined that the laser's beam was partially linearly polarized and the graph of the light intensity read by the LDR as a function of the angle was sinusoidal with minima when the filter was orthogonal to the direction of the partial polarization of the beam. Brewster's angle was calculated to be 56°, which is approximately the theoretical value of 56.3°, and the intensity distribution bell curve was fairly close to the expected shape with a slight shift to one side. 

## Introduction

A linear polarizer limits an electric field incident on it to just one plane. When a beam that is already linearly polarized is incident on a second linearly polarizing filter that is in the same orientation as the first, the transmitted beam will have the same power after both filters. However, if the second filter is placed at an angle $\theta$ relative to the first, the transmitted beam will have intensity $I$ calculated with the equation $$I = I_0 cos^2(\theta). \tag{1}$$ This relationship is evident when plotting the intensity as a function of the angle, as the maxima and minima range from 0 to $I$ sinusoidally.

Brewster's angle is the angle of incidence at which, when unpolarized light hits a barrier between two media, the reflected ray is perfectly polarized and the transmitted ray is partially polarized. When p-polarized light is incident on a barrier, the light is entirely transmitted; none of it is reflected. This is due to a phenomenon where the only light reflected from the medium is s-polarized because the reflected and refracted rays are 90° apart. At Brewster's angle, the electric field of the p-polarized light, which oscillates perpendicular to the wave's direction of propagation, would need to oscillate in the direction of the transmitted ray, whereas the electric field of the s-polarized light oscillates perpendicular to the angle of incidence and is reflected. Thus, all of the light is focused into the medium when the incident wave is p-polarized. The determination of Brewster's angle $\theta_B$ is dependent entirely on the indices of refraction of the two media on either side of the barrier. This makes the theoretical value for the expected angle of incidence at which there is no reflected light can be easily calculated using $$\theta_{B} = \frac{n_{1}}{n_{2}}, \tag{2}$$ where $n_1$ is the index of refraction of the first medium and $n_2$ the index of refraction of the medium the laser beam is propogating into. 

For the final portion wherein the Gaussian properties of the laser beam were investigated, specifically, the continuous probability distribution used to predict trends and properties of statistical models and, more relevantly, the intensity profile of a focused light source. A sharp edge is placed on a translational stage in front of a beam and is moved perpendicular to the beam's axis to slowly block more of the beam. As more of the beam is covered, the plot of the intensity distribution is an error function, $$erf(x) = \frac{2}{\sqrt{\pi}}\int_0^x e^{-t^2} dt , \tag{3}$$, for the derivative of which is the bell curve defined by $$f(x) = e^{-x^2}. \tag{4}$$ This function dictates that the center of the beam is the brightest and the intensity radiates from there and diminishes proportionally to distance from the center.
